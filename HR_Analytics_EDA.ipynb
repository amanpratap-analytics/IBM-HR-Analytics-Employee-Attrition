```python
categorical_cols = df.select_dtypes(include='object').columns

for col in categorical_cols:
    print(f"\n{'='*50}")
    print(df[col].value_counts())
```

```python
df.drop(
    ["EmployeeCount", "Over18", "StandardHours"],
    axis=1
)
```

```python
total_employee = len(df)
employees_left = (df["Attrition"] == 'Yes').sum()
employees_stayed = (df["Attrition"] == 'No').sum()
attrition_rate = employees_left / total_employee * 100

print(f"Total Employee: ", total_employee)
print(f"Employees Left: ", employees_left)
print(f"Employees Stayed: ", employees_stayed)
print(f"Attriton Rate: ", round(attrition_rate, 2), "%")
```

```python
df["Attrition"].value_counts()
```

```python
attrition_by_department = (
    df.groupby("Department")["Attrition"]
    .apply(lambda x: (x == "Yes").mean()*100)
    .sort_values(ascending=False)
)
attrition_by_department
```

```python
attrition_by_jobrole = (
    df.groupby("JobRole")["Attrition"]
    .apply(lambda x: (x == "Yes").mean()*100)
    .sort_values(ascending=False)
)
attrition_by_jobrole
```
